# **📘 Machine Learning — Mathematical Interview Q&A**

---

## **1. What is the hypothesis function in Linear Regression?**

The hypothesis (prediction) function models a linear relationship between features $x$ and output $y$:

$$
\hat{y} = h_\theta(x) = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \dots + \theta_n x_n
$$

or in vectorized form:

$$
h_\theta(x) = \theta^T x
$$

where

* $\theta$ = parameter vector (weights),
* $x$ = input feature vector.

---

## **2. What is the cost function for Linear Regression?**

We use **Mean Squared Error (MSE)**:

$$
J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right)^2
$$

* $m$ = number of training samples
* The factor $\frac{1}{2}$ simplifies gradient expressions.

---

## **3. Derive the Gradient Descent Update Rule for Linear Regression**

To minimize $J(\theta)$, update $\theta$ iteratively:

$$
\theta_j := \theta_j - \alpha \frac{\partial J(\theta)}{\partial \theta_j}
$$

Gradient for parameter $\theta_j$:

$$
\frac{\partial J(\theta)}{\partial \theta_j} = \frac{1}{m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right) x_j^{(i)}
$$

Thus:

$$
\boxed{
\theta_j := \theta_j - \frac{\alpha}{m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right) x_j^{(i)}
}
$$

where $\alpha$ is the **learning rate**.

---

## **4. What is the hypothesis function in Logistic Regression?**

Unlike Linear Regression, Logistic Regression outputs probabilities via the **sigmoid** (logistic) function:

$$
h_\theta(x) = \sigma(\theta^T x)
$$

where

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

and
$0 < h_\theta(x) < 1$

---

## **5. What is the cost function for Logistic Regression?**

We use **Binary Cross-Entropy (Log Loss):**

$$
J(\theta) = -\frac{1}{m} \sum_{i=1}^{m} \left[
y^{(i)} \log(h_\theta(x^{(i)})) +
(1 - y^{(i)}) \log(1 - h_\theta(x^{(i)}))
\right]
$$

This penalizes incorrect confident predictions heavily.

---

## **6. Derive the Gradient for Logistic Regression**

The gradient of $J(\theta)$ is:

$$
\frac{\partial J(\theta)}{\partial \theta_j} =
\frac{1}{m} \sum_{i=1}^{m}
\left( h_\theta(x^{(i)}) - y^{(i)} \right) x_j^{(i)}
$$

Thus, the **update rule** is identical in form to Linear Regression:

$$
\boxed{
\theta := \theta - \alpha \frac{1}{m} X^T (h_\theta(X) - y)
}
$$

---

## **7. What is the Difference Between MSE and Cross-Entropy Loss?**

| Aspect              | Mean Squared Error (MSE)           | Cross-Entropy Loss                                         |
| ------------------- | ---------------------------------- | ---------------------------------------------------------- |
| Type                | Regression tasks                   | Classification tasks                                       |
| Formula             | $\frac{1}{2m}\sum (y - \hat{y})^2$ | $-\frac{1}{m}\sum [y\log(\hat{y}) + (1-y)\log(1-\hat{y})]$ |
| Gradient Smoothness | Linear                             | Non-linear (sigmoid/softmax involved)                      |

## **8. What is Gradient Descent Intuitively?**

Gradient Descent is an iterative optimization algorithm to minimize a cost function $J(\theta)$.

Update rule:

$$
\theta := \theta - \alpha \nabla_\theta J(\theta)
$$

where

* $\alpha$ = learning rate
* $\nabla_\theta J(\theta)$ = gradient vector (partial derivatives)

It moves $\theta$ in the direction of **steepest descent** of $J(\theta)$.

---

## **9. What are Batch, Stochastic, and Mini-Batch Gradient Descent?**

| Type                    | Description                     | Formula                                                            |
| ----------------------- | ------------------------------- | ------------------------------------------------------------------ |
| **Batch GD**            | Uses all training examples      | $\theta := \theta - \alpha \frac{1}{m} X^T(h_\theta(X) - y)$       |
| **Stochastic GD (SGD)** | Uses one example at a time      | $\theta := \theta - \alpha (h_\theta(x^{(i)}) - y^{(i)})x^{(i)}$   |
| **Mini-Batch GD**       | Uses subset (batch) of examples | $\theta := \theta - \alpha \frac{1}{b} X_b^T(h_\theta(X_b) - y_b)$ |

## **10. What is Regularization, and Why is it Used?**

Regularization prevents **overfitting** by adding a penalty term to the cost function.

### **L2 (Ridge) Regularization:**

$$
J(\theta) = \frac{1}{2m}\sum (h_\theta(x^{(i)}) - y^{(i)})^2 + \frac{\lambda}{2m} \sum_{j=1}^{n} \theta_j^2
$$

### **L1 (Lasso) Regularization:**

$$
J(\theta) = \frac{1}{2m}\sum (h_\theta(x^{(i)}) - y^{(i)})^2 + \frac{\lambda}{m} \sum_{j=1}^{n} |\theta_j|
$$

where $\lambda$ controls regularization strength.

---

## **11. What is the Gradient of the L2 Regularized Cost Function?**

$$
\frac{\partial J(\theta)}{\partial \theta_j} =
\frac{1}{m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})x_j^{(i)} + \frac{\lambda}{m}\theta_j
$$

Update Rule:

$$
\boxed{
\theta_j := (1 - \alpha \frac{\lambda}{m})\theta_j - \alpha \frac{1}{m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})x_j^{(i)}
}
$$

---

## **12. What is the Sigmoid Derivative?**

If $\sigma(z) = \frac{1}{1+e^{-z}}$, then:

$$
\sigma'(z) = \sigma(z) (1 - \sigma(z))
$$

This property simplifies gradient calculations in logistic regression and neural networks.

---

## **13. What is the Softmax Function?**

Used in **multi-class** classification to convert logits into probabilities:

$$
P(y=i|x) = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}
$$

where $K$ = number of classes.

---

## **14. What is Cross-Entropy Loss for Multi-Class Classification?**

$$
J(\theta) = -\frac{1}{m} \sum_{i=1}^{m} \sum_{k=1}^{K} y_k^{(i)} \log(\hat{y}_k^{(i)})
$$

where

* $y_k^{(i)}$ = 1 if class $k$ is correct, else 0
* $\hat{y}_k^{(i)}$ = softmax output for class $k$

---

## **15. What is the Difference Between L1 and L2 Regularization?**

| Property | L1 (Lasso)                   | L2 (Ridge)                     |   |                           |
| -------- | ---------------------------- | ------------------------------ | - | ------------------------- |
| Penalty  | $\lambda \sum                | \theta_j                       | $ | $\lambda \sum \theta_j^2$ |
| Effect   | Feature selection (sparsity) | Weight shrinkage               |   |                           |
| Gradient | Non-differentiable at 0      | Smooth & differentiable        |   |                           |
| Output   | Sparse coefficients          | Small but nonzero coefficients |   |                           |

## **16. What is the Gradient of the Mean Squared Error (Vectorized)?**

Let $X$ be $(m \times n)$, $y$ be $(m \times 1)$, $\theta$ be $(n \times 1)$:

$$
\nabla_\theta J(\theta) = \frac{1}{m} X^T (X\theta - y)
$$

## **17. What are Evaluation Metrics for Classification Models?**

| Metric                   | Formula                                                       | Meaning                                  |
| ------------------------ | ------------------------------------------------------------- | ---------------------------------------- |
| **Accuracy**             | $\frac{TP + TN}{TP + TN + FP + FN}$                           | Overall correctness                      |
| **Precision**            | $\frac{TP}{TP + FP}$                                          | How many predicted positives are correct |
| **Recall (Sensitivity)** | $\frac{TP}{TP + FN}$                                          | How many actual positives are caught     |
| **F1-Score**             | $2 \times \frac{Precision \times Recall}{Precision + Recall}$ | Harmonic mean of precision and recall    |

## **18. What is the Confusion Matrix?**

| Actual \ Predicted | Positive | Negative |
| ------------------ | -------- | -------- |
| **Positive**       | TP       | FN       |
| **Negative**       | FP       | TN       |

Where:

* TP = True Positive
* TN = True Negative
* FP = False Positive
* FN = False Negative

## **19. What is Feature Scaling and Why Is It Important?**

Feature scaling ensures all features contribute equally to gradient descent.

### **Standardization:**

$$
x' = \frac{x - \mu}{\sigma}
$$

### **Min-Max Normalization:**

$$
x' = \frac{x - x_{min}}{x_{max} - x_{min}}
$$

Unscaled data → slow or unstable convergence in gradient descent.

## **20. What is the Difference Between Convex and Non-Convex Functions?**

A function $f(x)$ is **convex** if:

$$
f(\lambda x_1 + (1 - \lambda)x_2) \leq \lambda f(x_1) + (1 - \lambda) f(x_2), \quad \forall \lambda \in [0,1]
$$

* **Linear Regression loss** is convex → unique global minimum
* **Neural Network loss** is non-convex → many local minima
